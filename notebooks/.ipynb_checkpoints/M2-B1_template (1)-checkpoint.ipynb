```python
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd

def top_mots_par_classe(
    dataframe,
    classe,
    text_col="synthese_entretien",
    target_col="classe_retour_emploi",
    top_n=15,
):
    
    corpus = (
        dataframe.loc[
            dataframe[target_col] == classe,
            text_col
        ]
        .fillna("")
        .astype(str)
    )

    vectorizer = CountVectorizer(
        stop_words=stop_words_list,
        max_features=1000,
        ngram_range=(2,3)
    )

    X = vectorizer.fit_transform(corpus)

    freq = pd.DataFrame({
        "mot": vectorizer.get_feature_names_out(),
        "frequence": X.sum(axis=0).A1
    })

    return (
        freq.sort_values(
            "frequence",
            ascending=False
        )
        .head(top_n)
    )
```

```python
def compute_disparate_impact(df, sensitive_col, target_col, positive_value):

    selection_rate = (
        df.groupby(sensitive_col, observed=True)[target_col]
          .apply(lambda x: (x == positive_value).mean())
    )

    sr_min = selection_rate.min()
    sr_max = selection_rate.max()
    di = sr_min / sr_max

    if di < 0.8:
        verdict = "⚠️ Signal"
    elif di > 1.25:
        verdict = "⚠️ Signal inversé"
    else:
        verdict = "✅ OK"

    return {
        "Variable sensible": sensitive_col,
        "Selection Rate min": round(sr_min, 3),
        "Selection Rate max": round(sr_max, 3),
        "DI": round(di, 3),
        "Verdict": verdict
    }
```

```python
import pandas as pd
import numpy as np

def compute_disparate_impact(
    df,
    sensitive_col,
    favorable_class=0,
):
    """
    Calcule le Disparate Impact pour une variable binaire.

    DI = taux favorable groupe 1 / taux favorable groupe 0
    """

    data = df[
        [sensitive_col, "classe_retour_emploi"]
    ].dropna()

    values = sorted(
        data[sensitive_col].unique()
    )

    if len(values) != 2:
        return None

    g0, g1 = values

    rate0 = (
        data.loc[
            data[sensitive_col] == g0,
            "classe_retour_emploi"
        ]
        .eq(favorable_class)
        .mean()
    )

    rate1 = (
        data.loc[
            data[sensitive_col] == g1,
            "classe_retour_emploi"
        ]
        .eq(favorable_class)
        .mean()
    )

    di = np.nan

    if rate0 > 0:
        di = rate1 / rate0

    return {
        "variable": sensitive_col,
        "groupe_reference": g0,
        "groupe_compare": g1,
        "taux_ref": round(rate0, 3),
        "taux_compare": round(rate1, 3),
        "DI": round(di, 3),
    }
```

```python
df["age_50+"] = (
    df["age"] >= 50
).astype(int)
```

```python
df["diplome_sup"] = (
    df["niveau_diplome"]
    .isin([
        "Bac+2",
        "Bac+5"
    ])
).astype(int)
```

```python
variables_auditer = [
    "nationalite_hors_ue",
    "est_allocataire",
    "age_50+",
    "diplome_sup",
]

results = []

for col in variables_auditer:

    res = compute_disparate_impact(
        df,
        sensitive_col=col,
        favorable_class=0
    )

    if res is not None:
        results.append(res)

audit_di = pd.DataFrame(results)

display(audit_di)
```

```python
def interpret_di(di):

    if pd.isna(di):
        return "NA"

    if di < 0.80:
        return "Risque de biais"

    if di <= 1.25:
        return "Acceptable"

    return "A surveiller"
```

```python
audit_di["Interpretation"] = (
    audit_di["DI"]
    .apply(interpret_di)
)

display(audit_di)
```

### 🎓 Pourquoi un `Pipeline` ? (définition — la fuite de données)

Tu pourrais imputer/encoder/scaler « à la main » en Pandas. On enferme tout
dans un `Pipeline` parce qu'un préprocesseur **apprend des paramètres sur les
données** : `SimpleImputer` retient une médiane, `StandardScaler` une moyenne
et un écart-type, `OneHotEncoder` la liste des modalités.

**Fuite de données** (*data leakage*) = laisser de l'information du jeu de
**test** influencer la préparation du jeu d'**entraînement** — typiquement
calculer ces paramètres sur *tout* le dataset avant de le splitter. Symptôme :
des scores de validation **anormalement optimistes** qui s'effondrent en prod.

Le `Pipeline` est l'outil qui rend l'étanchéité automatique : un seul `fit` sur
le train propage la discipline à toutes les transformations. **Ici tu n'as pas
de split** (on produit juste un dataset propre) → `fit_transform` sur tout le
dataset est OK. **Réflexe à garder** : *dès qu'un modèle entre en jeu, le
préprocesseur ne se `fit` que sur le train.* → mise en œuvre en **M4-B1**.

```python
import sys
sys.path.append("../src")
from preprocess import build_preprocessor, load_dataset

X, y = load_dataset(RAW_PATH)
preprocessor = build_preprocessor()
X_transformed = preprocessor.fit_transform(X)
print(f"Shape originale : {X.shape}")
print(f"Shape transformée : {X_transformed.shape}")
```
