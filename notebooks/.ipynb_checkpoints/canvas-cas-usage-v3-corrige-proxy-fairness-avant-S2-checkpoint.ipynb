- `0` : retour rapide, < 6 mois ;

Le sujet demande notamment plusieurs modèles, Accuracy, F1 macro, matrice de confusion, validation croisée stratifiée, imputation, feature engineering géographique et comparaison de scénarios. fileciteturn0file0L43-L60

### 3.7 Équité : ne pas confondre association et discrimination

Une différence de taux entre groupes **ne prouve pas** à elle seule une discrimination algorithmique. Elle signale un écart à investiguer.

Pour l’audit, on mesure :
1. le taux de classe 0 (« issue favorable ») via un **Disparate Impact** indicatif ;
2. surtout, le **rappel de la classe 2 par groupe**, car l’enjeu métier porte sur la détection des situations à risque ;
3. après entraînement, les mêmes indicateurs sur les **prédictions**.

Le seuil de 0,80 du ratio des taux est une heuristique d’audit, pas une preuve juridique.


````python
# 3.8 Audit initial des écarts de cible
def disparate_impact(data, group_col, target_col="classe_retour_emploi", positive_class=0, min_group_size=15):
    tmp = data[[group_col, target_col]].dropna().copy()
    rates = tmp.groupby(group_col, observed=True).agg(
        n=(target_col, "size"),
        selection_rate=(target_col, lambda s: (s == positive_class).mean())
    )
    rates = rates[rates["n"] >= min_group_size]
    if len(rates) < 2:
        return np.nan, rates
    return rates["selection_rate"].min() / rates["selection_rate"].max(), rates.sort_values("selection_rate")

for col in ["nationalite_hors_ue", "groupe_age", "region"]:
    di, rates = disparate_impact(df, col)
    print(f"\n{col} — DI classe 0 (n≥15) : {di:.3f}")
    display(rates)

print("\nRépartition de la classe 2 par groupe :")
for col in ["nationalite_hors_ue", "groupe_age", "region"]:
    tmp = pd.crosstab(df[col], df["classe_retour_emploi"], normalize="index")
    print(f"\n{col}")
    display(tmp.round(3))
````

## Analyse des variables proxy et fairness

Avant de construire le scénario S2, nous vérifions si des variables restantes peuvent contenir indirectement de l'information liée à `nationalite_hors_ue`.

**Principe :**
- `nationalite_hors_ue` reste disponible ici uniquement comme variable de référence pour l'audit ;
- elle sera supprimée des variables explicatives du scénario S2 ;
- une association statistique n'est **pas** une preuve de discrimination ;
- nous complétons l'analyse par un modèle auxiliaire permettant de mesurer la capacité des variables candidates à prédire la variable sensible.

Variables examinées : géographie/INSEE, diplôme, ROME, allocation, âge, ancienneté et texte d'entretien.

````python
# Imports nécessaires à l'analyse proxy
from scipy.stats import chi2_contingency
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

proxy_required = [
    "nationalite_hors_ue", "age", "niveau_diplome",
    "anciennete_poste_ans", "code_rome_vise",
    "code_insee_commune", "est_allocataire",
    "synthese_entretien"
]
missing_proxy_cols = [c for c in proxy_required if c not in df.columns]
assert not missing_proxy_cols, f"Colonnes manquantes : {missing_proxy_cols}"

print("Analyse proxy prête :", len(proxy_required), "variables contrôlées.")
````

### 1. Associations avec la variable sensible

````python
categorical_proxy_vars = [
    "niveau_diplome",
    "code_rome_vise",
    "code_insee_commune",
    "est_allocataire"
]

proxy_chi2_results = []

for col in categorical_proxy_vars:
    tmp = df[[col, "nationalite_hors_ue"]].dropna()
    table = pd.crosstab(tmp[col], tmp["nationalite_hors_ue"])

    if table.shape[0] > 1 and table.shape[1] > 1:
        chi2, p_value, dof, expected = chi2_contingency(table)
        proxy_chi2_results.append({
            "variable": col,
            "n_modalites": table.shape[0],
            "chi2": chi2,
            "p_value": p_value
        })

proxy_chi2_results = pd.DataFrame(proxy_chi2_results).sort_values("p_value")
display(proxy_chi2_results)
````

**Interprétation :** une p-value faible indique une association statistique. Elle ne permet pas, à elle seule, de conclure qu'une variable est discriminatoire ou qu'elle doit être supprimée.

### 2. Analyse géographique : extraction du département

````python
def extract_department(code):
    if pd.isna(code):
        return np.nan
    code = str(code).strip().upper().zfill(5)
    if code.startswith(("2A", "2B")):
        return code[:2]
    return code[:2]

df["departement"] = df["code_insee_commune"].apply(extract_department)

geo_proxy_table = pd.crosstab(
    df["departement"],
    df["nationalite_hors_ue"],
    normalize="index"
)

display(geo_proxy_table.head(20))

geo_rate = (
    df.groupby("departement")["nationalite_hors_ue"]
      .agg(["mean", "count"])
      .rename(columns={"mean": "taux_hors_ue"})
      .sort_values("taux_hors_ue", ascending=False)
)

display(geo_rate.head(15))
````

### 3. Test de proxy par modèle auxiliaire

Nous construisons temporairement un modèle dont la cible est `nationalite_hors_ue`.

Une AUC nettement supérieure à 0,50 signifie que les variables candidates contiennent un signal permettant de retrouver indirectement la variable sensible. C'est un **signal de proxy potentiel**, pas une preuve de discrimination.

````python
# Proxy tabulaire : variables candidates -> nationalite_hors_ue
X_proxy = df[
    [
        "age",
        "niveau_diplome",
        "anciennete_poste_ans",
        "code_rome_vise",
        "departement",
        "est_allocataire"
    ]
].copy()

y_proxy = df["nationalite_hors_ue"].copy()

num_proxy_cols = ["age", "anciennete_poste_ans"]
cat_proxy_cols = [
    "niveau_diplome",
    "code_rome_vise",
    "departement",
    "est_allocataire"
]

proxy_preprocessor = ColumnTransformer([
    ("num", SimpleImputer(strategy="median"), num_proxy_cols),
    (
        "cat",
        Pipeline([
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("onehot", OneHotEncoder(handle_unknown="ignore"))
        ]),
        cat_proxy_cols
    )
])

proxy_model = Pipeline([
    ("preprocessing", proxy_preprocessor),
    ("model", RandomForestClassifier(
        n_estimators=300,
        random_state=42,
        class_weight="balanced",
        n_jobs=-1
    ))
])

proxy_cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

proxy_auc_scores = cross_val_score(
    proxy_model,
    X_proxy,
    y_proxy,
    cv=proxy_cv,
    scoring="roc_auc",
    n_jobs=-1
)

print(f"AUC proxy tabulaire : {proxy_auc_scores.mean():.3f} ± {proxy_auc_scores.std():.3f}")
````

````python
# Proxy texte : synthese_entretien -> nationalite_hors_ue
X_text_proxy = df["synthese_entretien"].fillna("").astype(str)
y_text_proxy = df["nationalite_hors_ue"]

text_proxy_model = Pipeline([
    ("tfidf", TfidfVectorizer(
        lowercase=True,
        ngram_range=(1, 2),
        min_df=2
    )),
    ("model", LogisticRegression(
        max_iter=1000,
        class_weight="balanced"
    ))
])

text_proxy_auc_scores = cross_val_score(
    text_proxy_model,
    X_text_proxy,
    y_text_proxy,
    cv=proxy_cv,
    scoring="roc_auc",
    n_jobs=-1
)

print(f"AUC proxy texte : {text_proxy_auc_scores.mean():.3f} ± {text_proxy_auc_scores.std():.3f}")
````

### 4. Inspection exploratoire du texte

````python
sensitive_terms = [
    "nationalité", "nationalite",
    "origine", "pays",
    "langue", "linguistique",
    "immigration", "immigré", "immigre",
    "titre de séjour", "titre de sejour"
]

text_series = df["synthese_entretien"].fillna("").astype(str).str.lower()

term_counts = pd.DataFrame([
    {
        "terme": term,
        "occurrences": int(text_series.str.contains(term, regex=False).sum())
    }
    for term in sensitive_terms
]).sort_values("occurrences", ascending=False)

display(term_counts)
````

### 5. Conclusion fairness / proxy

- Une corrélation avec `nationalite_hors_ue` n'est pas automatiquement une discrimination.
- Une capacité prédictive élevée de la variable sensible indique un **risque de proxy potentiel**.
- Le maintien d'une variable doit être justifié par sa pertinence métier.
- Le scénario S2 supprime explicitement `nationalite_hors_ue`, puis mesure les performances et les disparités restantes.
- L'analyse finale doit notamment suivre l'erreur critique **classe réelle 2 → classe prédite 0**.

Cette approche distingue **corrélation statistique**, **risque de proxy** et **impact fairness réel**.

````python
# Décision opérationnelle issue de l'analyse proxy
tab_auc_mean = float(np.mean(proxy_auc_scores))
text_auc_mean = float(np.mean(text_proxy_auc_scores))

department_p = np.nan
if "departement" in proxy_chi2_results["variable"].values:
    department_p = float(
        proxy_chi2_results.loc[
            proxy_chi2_results["variable"].eq("departement"), "p_value"
        ].iloc[0]
    )

sensitive_text_hits = int(term_counts["occurrences"].sum())

# Seuil exploratoire : AUC >= 0.60 = signal prédictif notable.
# Il ne s'agit pas d'un seuil réglementaire ni d'une preuve de discrimination.
proxy_flags = []

if tab_auc_mean >= 0.60:
    proxy_flags.append("signal tabulaire")
if text_auc_mean >= 0.60:
    proxy_flags.append("signal texte")
if (
    not np.isnan(department_p)
    and department_p < 0.05
    and tab_auc_mean >= 0.60
):
    proxy_flags.append("signal géographique")

# Mitigation automatique uniquement si le signal est à la fois substantiel
# et cohérent avec l'analyse statistique.
PROXY_MITIGATION_FEATURES = []
if (
    tab_auc_mean >= 0.60
    and not np.isnan(department_p)
    and department_p < 0.05
):
    PROXY_MITIGATION_FEATURES.append("region")

proxy_risk_level = (
    "À surveiller / mitigation possible"
    if proxy_flags
    else "Faible à modéré — pas de mitigation automatique"
)

print(f"AUC proxy tabulaire : {tab_auc_mean:.3f}")
print(f"AUC proxy texte      : {text_auc_mean:.3f}")
print(
    f"p-value département  : {department_p:.4f}"
    if not np.isnan(department_p)
    else "p-value département : non disponible"
)
print(f"Occurrences de termes sensibles dans le texte : {sensitive_text_hits}")
print("Signal(s) retenu(s) :", proxy_flags or "aucun signal substantiel")
print("Niveau de risque proxy :", proxy_risk_level)
print("Mitigation appliquée à S2 :", PROXY_MITIGATION_FEATURES or "aucune")

if not proxy_flags:
    print("\nConclusion opérationnelle :")
    print("- les signaux de proxy sont faibles/modérés ;")
    print("- S2 retire la variable sensible explicite ;")
    print("- les variables métier restantes sont conservées ;")
    print("- la fairness réelle sera vérifiée après entraînement.")
else:
    print("\nConclusion opérationnelle : un signal proxy substantiel a été détecté.")
    print("S2 applique donc la mitigation définie ci-dessus, puis l'impact est contrôlé sur les performances et la fairness.")
````

## 4. Préparation et scénarios

**Règle anti-fuite :** toutes les imputations, standardisations, encodages et vectorisations TF-IDF sont apprises **à l’intérieur du pipeline**, donc séparément dans chaque fold de CV.

### Scénarios demandés
- **S1 — multimodal complet** : tabulaire + TF-IDF + variable sensible ;
- **S2 — approche éthique** : tabulaire + TF-IDF, sans `nationalite_hors_ue` ;
- **S3 — texte seul** : TF-IDF uniquement ;
- **S4 — tabulaire seul** : variables tabulaires, sans texte.

### Décision issue de l'analyse proxy/fairness

L'analyse proxy est volontairement exécutée **avant S2**.

**Principe de décision :**
1. `nationalite_hors_ue` est retirée de S2 conformément au sujet ;
2. une corrélation avec la variable sensible ne suffit pas à justifier la suppression d'une variable métier ;
3. une mitigation complémentaire est appliquée seulement si l'analyse met en évidence un signal proxy substantiel ;
4. l'impact réel est ensuite vérifié avec les métriques de performance et de fairness.

La variable `PROXY_MITIGATION_FEATURES` calculée ci-dessus est donc utilisée directement dans la définition de S2.


````python
# 4.1 Split unique train/test, stratifié
TARGET = "classe_retour_emploi"

feature_df = df.copy()
X = feature_df.drop(columns=[TARGET])
y = feature_df[TARGET]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, stratify=y, random_state=RANDOM_STATE
)

print("Train :", X_train.shape, y_train.value_counts(normalize=True).sort_index().round(3).to_dict())
print("Test  :", X_test.shape, y_test.value_counts(normalize=True).sort_index().round(3).to_dict())
````

````python
# 4.2 Préprocesseur réutilisable

NUM_FEATURES = ["age", "anciennete_poste_ans"]
BASE_CAT_FEATURES = ["niveau_diplome", "code_rome_famille", "region", "est_allocataire"]
SENSITIVE_FEATURE = "nationalite_hors_ue"
TEXT_FEATURE = "synthese_entretien"

# Décision de mitigation issue de l'analyse proxy précédente.
MITIGATION_CAT_FEATURES = [c for c in PROXY_MITIGATION_FEATURES if c in BASE_CAT_FEATURES]

def make_preprocessor(num_features, cat_features, use_text):
    transformers = []

    if num_features:
        num_pipe = Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("scaler", StandardScaler())
        ])
        transformers.append(("num", num_pipe, num_features))

    if cat_features:
        cat_pipe = Pipeline([
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("onehot", OneHotEncoder(handle_unknown="ignore"))
        ])
        transformers.append(("cat", cat_pipe, cat_features))

    if use_text:
        text_pipe = Pipeline([
            ("imputer", SimpleImputer(strategy="constant", fill_value="")),
            ("flatten", FunctionTransformer(lambda x: x.ravel(), validate=False)),
            ("tfidf", TfidfVectorizer(
                lowercase=True,
                strip_accents="unicode",
                ngram_range=(1, 2),
                min_df=2,
                max_features=3000
            ))
        ])
        transformers.append(("text", text_pipe, [TEXT_FEATURE]))

    if not transformers:
        raise ValueError("Aucune feature sélectionnée")

    return ColumnTransformer(transformers, remainder="drop")

scenarios = {
    "S1": {
        "num": NUM_FEATURES,
        "cat": BASE_CAT_FEATURES + [SENSITIVE_FEATURE],
        "text": True,
        "description": "Multimodal complet"
    },
    "S2": {
        "num": NUM_FEATURES,
        "cat": [c for c in BASE_CAT_FEATURES if c not in MITIGATION_CAT_FEATURES],
        "text": True,
        "description": "Sans variable sensible + mitigation des proxies identifiés"
    },
    "S3": {
        "num": [],
        "cat": [],
        "text": True,
        "description": "Texte seul (TF-IDF)"
    },
    "S4": {
        "num": NUM_FEATURES,
        "cat": BASE_CAT_FEATURES + [SENSITIVE_FEATURE],
        "text": False,
        "description": "Tabulaire seul"
    }
}

print("Variables retirées de S2 au titre de la mitigation proxy :", MITIGATION_CAT_FEATURES or "Aucune")
pd.DataFrame(scenarios).T
````

````python
# 4.3 Assertions
assert len(X_train) + len(X_test) == len(df)
assert y_train.notna().all() and y_test.notna().all()
assert set(y_train.unique()) == {0,1,2}
assert set(y_test.unique()) == {0,1,2}
assert X_train["usager_id"].is_unique
assert X_test["usager_id"].is_unique

# L'identifiant et le code commune brut sont volontairement exclus des features.
assert "usager_id" not in NUM_FEATURES + CAT_FEATURES
assert "code_insee_commune" not in NUM_FEATURES + CAT_FEATURES

print("Assertions de qualité : OK")
````

## 5. Modélisation

### Familles considérées
- **ML classique** : retenu, car le problème est une classification tabulaire de petite taille avec variables catégorielles et texte court.
- **Deep Learning** : écarté comme solution principale ; 2 500 lignes et 9 formulations textuelles ne justifient pas le coût et la complexité.
- **SLM / LLM / RAG / agentique** : hors besoin métier ; ils ne sont pas nécessaires pour produire une classe structurée et augmenteraient le coût d’inférence et les contraintes de gouvernance.

Le sujet demande surtout d’argumenter le choix de famille, pas d’implémenter toutes les familles. fileciteturn0file1L1-L7


````python
# 5.1 Modèles candidats
models = {
    "LogReg": LogisticRegression(
        max_iter=2000, class_weight="balanced", random_state=RANDOM_STATE
    ),
    "RandomForest": RandomForestClassifier(
        n_estimators=150, class_weight="balanced",
        min_samples_leaf=2, random_state=RANDOM_STATE, n_jobs=-1
    ),
    "LightGBM": LGBMClassifier(
        n_estimators=150, learning_rate=0.05, num_leaves=31,
        class_weight="balanced", random_state=RANDOM_STATE, verbosity=-1
    ),
    "XGBoost": XGBClassifier(
        n_estimators=150, max_depth=5, learning_rate=0.05,
        subsample=0.9, colsample_bytree=0.9,
        objective="multi:softprob", eval_metric="mlogloss",
        random_state=RANDOM_STATE, n_jobs=4
    )
}

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)

def recall_class2(y_true, y_pred):
    return recall_score(y_true, y_pred, labels=[2], average=None, zero_division=0)[0]

def critical_20(y_true, y_pred):
    cm = confusion_matrix(y_true, y_pred, labels=[0,1,2])
    return cm[2, 0] / cm[2].sum() if cm[2].sum() else np.nan

scoring = {
    "accuracy": "accuracy",
    "f1_macro": "f1_macro",
    "recall_class2": make_scorer(recall_class2),
    "critical_2_to_0": make_scorer(critical_20)
}
````

````python
# 5.2 CV : comparaison des scénarios et modèles
cv_rows = []

for scenario_name, spec in scenarios.items():
    prep = make_preprocessor(spec["num"], spec["cat"], spec["text"])

    for model_name, model in models.items():
        pipe = Pipeline([
            ("prep", prep),
            ("clf", model)
        ])

        start = time.perf_counter()
        scores = cross_validate(
            pipe, X_train, y_train,
            cv=cv,
            scoring=scoring,
            n_jobs=1,
            return_train_score=False,
            error_score="raise"
        )
        elapsed = time.perf_counter() - start

        cv_rows.append({
            "scenario": scenario_name,
            "model": model_name,
            "accuracy_cv": scores["test_accuracy"].mean(),
            "f1_macro_cv": scores["test_f1_macro"].mean(),
            "recall_classe_2_cv": scores["test_recall_class2"].mean(),
            "erreur_2_vers_0_cv": scores["test_critical_2_to_0"].mean(),
            "f1_std": scores["test_f1_macro"].std(),
            "temps_cv_s": elapsed
        })

cv_results = pd.DataFrame(cv_rows).sort_values(
    ["f1_macro_cv", "recall_classe_2_cv"], ascending=False
).reset_index(drop=True)

display(cv_results.style.format({
    "accuracy_cv":"{:.3f}",
    "f1_macro_cv":"{:.3f}",
    "recall_classe_2_cv":"{:.3f}",
    "erreur_2_vers_0_cv":"{:.3f}",
    "f1_std":"{:.3f}",
    "temps_cv_s":"{:.1f}"
}))
````

````python
# 5.3 Visualisation du compromis performance / erreur critique
plt.figure(figsize=(9,6))
for scenario_name, g in cv_results.groupby("scenario"):
    plt.scatter(g["recall_classe_2_cv"], g["f1_macro_cv"], label=scenario_name)
plt.axvline(0.75, linestyle="--", label="Repère rappel classe 2 = 0,75")
plt.xlabel("Rappel CV — classe 2")
plt.ylabel("F1 macro CV")
plt.title("Compromis F1 macro / détection du risque de longue durée")
plt.legend(bbox_to_anchor=(1.02,1), loc="upper left")
plt.tight_layout()
plt.show()
````

### 5.4 Sélection du candidat

La sélection ne repose pas uniquement sur le meilleur F1 :
1. F1 macro et Accuracy pour la performance globale ;
2. rappel de la classe 2 pour ne pas négliger le risque de longue durée ;
3. taux d’erreur critique 2→0 ;
4. temps de calcul et complexité d’inférence.

Le **test set reste intact** jusqu’à cette étape. Le choix se fait exclusivement sur le train/CV.


````python
# 5.5 Sélection automatique d'un candidat raisonnable
# Priorité : F1 macro élevé, puis rappel classe 2, avec pénalité si erreur critique très élevée.
selection = cv_results.copy()
selection["score_decision"] = (
    selection["f1_macro_cv"]
    + 0.25 * selection["recall_classe_2_cv"]
    - 0.50 * selection["erreur_2_vers_0_cv"]
)

best_row = selection.sort_values("score_decision", ascending=False).iloc[0]
BEST_SCENARIO = best_row["scenario"]
BEST_MODEL_NAME = best_row["model"]

print("Candidat retenu pour l'optimisation finale :")
print(BEST_SCENARIO, BEST_MODEL_NAME)
display(best_row.to_frame("valeur"))
````

## 5.6 Optimisation légère des hyperparamètres

On ne cherche pas une optimisation exhaustive. Le but est de montrer que les paramètres influencent le compromis performance / risque et de rester reproductible.


````python
# 5.6 GridSearch compact sur le candidat retenu
best_spec = scenarios[BEST_SCENARIO]
best_prep = make_preprocessor(best_spec["num"], best_spec["cat"], best_spec["text"])

base_model = models[BEST_MODEL_NAME]

param_grids = {
    "LogReg": {
        "clf__C": [0.5, 1.0, 2.0],
        "clf__class_weight": ["balanced", {0:1, 1:1, 2:1.5}, {0:1, 1:1, 2:2}]
    },
    "RandomForest": {
        "clf__n_estimators": [100, 200],
        "clf__max_depth": [None, 12],
        "clf__min_samples_leaf": [1, 2],
        "clf__class_weight": ["balanced", {0:1,1:1,2:2}]
    },
    "LightGBM": {
        "clf__n_estimators": [100, 200],
        "clf__num_leaves": [15, 31],
        "clf__class_weight": ["balanced", {0:1,1:1,2:2}]
    },
    "XGBoost": {
        "clf__max_depth": [3, 5],
        "clf__n_estimators": [100, 200],
        "clf__learning_rate": [0.03, 0.05]
    }
}

grid = GridSearchCV(
    Pipeline([("prep", best_prep), ("clf", base_model)]),
    param_grid=param_grids[BEST_MODEL_NAME],
    scoring="f1_macro",
    cv=cv,
    n_jobs=1,
    refit=True
)

grid.fit(X_train, y_train)
best_pipe = grid.best_estimator_

print("Meilleurs paramètres :", grid.best_params_)
print("Meilleur F1 macro CV :", round(grid.best_score_, 3))
````

## 5.7 Évaluation finale sur le test

Le test est utilisé une seule fois pour annoncer le niveau de performance final. On affiche obligatoirement la matrice de confusion et l’erreur critique **2→0**.


````python
# 5.7 Test final
y_pred = best_pipe.predict(X_test)
cm = confusion_matrix(y_test, y_pred, labels=[0,1,2])

test_metrics = pd.Series({
    "Accuracy": accuracy_score(y_test, y_pred),
    "F1 macro": f1_score(y_test, y_pred, average="macro"),
    "Rappel classe 2": recall_class2(y_test, y_pred),
    "Erreur critique 2→0": critical_20(y_test, y_pred)
})

display(test_metrics.to_frame("score").style.format("{:.3f}"))

print(classification_report(
    y_test, y_pred, labels=[0,1,2],
    target_names=["0 — rapide", "1 — moyen", "2 — longue durée"],
    digits=3,
    zero_division=0
))

plt.figure(figsize=(5,4))
plt.imshow(cm, interpolation="nearest", aspect="auto")
for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):
        plt.text(j, i, cm[i, j], ha="center", va="center")
plt.xticks(range(3), [0,1,2])
plt.yticks(range(3), [0,1,2])
plt.xlabel("Prédiction")
plt.ylabel("Réel")
plt.title("Matrice de confusion — test")
plt.show()
````

## 5.8 Réduction des erreurs critiques par seuil de décision

Une pondération de classe agit pendant l’apprentissage. Un second levier consiste à modifier la règle de décision à partir des probabilités.

On choisit le seuil **uniquement sur les prédictions out-of-fold du train**, jamais sur le test. On cherche le seuil qui réduit 2→0 tout en conservant un F1 macro acceptable.


````python
# 5.8 Calibration opérationnelle du seuil classe 2 sur le train uniquement
oof_proba = cross_val_predict(
    best_pipe, X_train, y_train,
    cv=cv, method="predict_proba", n_jobs=1
)

def predict_with_class2_threshold(proba, threshold):
    pred = np.argmax(proba, axis=1)
    force_2 = proba[:, 2] >= threshold
    pred[force_2] = 2
    return pred

threshold_rows = []
for threshold in np.arange(0.15, 0.61, 0.02):
    pred = predict_with_class2_threshold(oof_proba, threshold)
    threshold_rows.append({
        "threshold_2": threshold,
        "f1_macro": f1_score(y_train, pred, average="macro"),
        "recall_class2": recall_class2(y_train, pred),
        "critical_2_to_0": critical_20(y_train, pred)
    })

threshold_df = pd.DataFrame(threshold_rows)

# Contraintes métier : priorité à la baisse de 2→0, sans sacrifier totalement le F1.
eligible = threshold_df[threshold_df["critical_2_to_0"] <= 0.10]
if len(eligible):
    chosen_threshold = eligible.sort_values(
        ["f1_macro", "recall_class2"], ascending=False
    ).iloc[0]["threshold_2"]
else:
    chosen_threshold = threshold_df.sort_values("critical_2_to_0").iloc[0]["threshold_2"]

print(f"Seuil classe 2 choisi sur OOF train : {chosen_threshold:.2f}")
display(threshold_df.sort_values("f1_macro", ascending=False).head(10))
````

````python
# 5.9 Test du seuil choisi — le test n'intervient qu'ici
test_proba = best_pipe.predict_proba(X_test)
y_pred_threshold = predict_with_class2_threshold(test_proba, chosen_threshold)

threshold_test_metrics = pd.Series({
    "Accuracy": accuracy_score(y_test, y_pred_threshold),
    "F1 macro": f1_score(y_test, y_pred_threshold, average="macro"),
    "Rappel classe 2": recall_class2(y_test, y_pred_threshold),
    "Erreur critique 2→0": critical_20(y_test, y_pred_threshold)
})

display(threshold_test_metrics.to_frame("score").style.format("{:.3f}"))

cm_thr = confusion_matrix(y_test, y_pred_threshold, labels=[0,1,2])
plt.figure(figsize=(5,4))
plt.imshow(cm_thr, interpolation="nearest", aspect="auto")
for i in range(cm_thr.shape[0]):
    for j in range(cm_thr.shape[1]):
        plt.text(j, i, cm_thr[i, j], ha="center", va="center")
plt.xticks(range(3), [0,1,2])
plt.yticks(range(3), [0,1,2])
plt.xlabel("Prédiction")
plt.ylabel("Réel")
plt.title("Matrice de confusion — test après seuil classe 2")
plt.show()
````

````python
# 6.2 Audit d'équité des prédictions finales

L'analyse proxy précédente identifie un **risque potentiel** ; cette étape mesure l'**impact réel des prédictions**.

On compare les groupes sur :
- taux de prédiction de la classe 0 ;
- rappel de la classe 2 ;
- erreur critique 2→0.

On vérifie ainsi si la suppression de `nationalite_hors_ue` et la mitigation éventuelle des proxies réduisent ou aggravent les disparités.
````

### 6.3 Recommandation

Le choix final doit privilégier un modèle qui :
- détecte correctement la classe 2 ;
- minimise les erreurs 2→0 ;
- reste suffisamment simple et rapide pour une utilisation en entretien ;
- limite les écarts injustifiés entre groupes ;
- conserve une traçabilité permettant une revue humaine.

**Important :** le modèle ne doit pas remplacer la décision du conseiller. Pour un cas à faible confiance ou à risque critique, l’architecture prévoit une abstention / escalade humaine.

Pour un modèle arborescent, utiliser les importances avec prudence ; pour une analyse plus robuste, SHAP peut être ajouté en complément. L’objectif est d’expliquer les facteurs utilisés par le modèle, pas de transformer une corrélation en causalité.
