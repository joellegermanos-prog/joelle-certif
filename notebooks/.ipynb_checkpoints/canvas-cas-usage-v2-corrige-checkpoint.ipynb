### 1.1 Contexte client

*[Décris en 5-10 lignes le client, son secteur, le problème métier qu'il rencontre, ce qui motive le recours à l'IA.]*
*Le client est une **agence publique nationale de l'emploi** (secteur des politiques publiques de l'emploi et de l'accompagnement social), chargée d'orienter et de suivre les demandeurs d'emploi tout au long de leur parcours. Concrètement, à chaque premier entretien d'aiguillage, un conseiller collecte des informations administratives et socio-professionnelles sur l'usager (âge, diplôme, ancienneté, statut d'allocataire, situation géographique) et rédige une synthèse libre décrivant le projet professionnel et les freins périphériques rencontrés (mobilité, garde d'enfants, compétences à actualiser, etc.).

Le problème métier est que ce diagnostic d'orientation repose aujourd'hui essentiellement sur l'appréciation humaine du conseiller, dans un temps d'entretien contraint, alors que l'enjeu est déterminant : repérer le plus tôt possible les usagers exposés à un **risque de chômage de longue durée (>12 mois)** pour leur proposer un accompagnement renforcé, sans pour autant sur-mobiliser des ressources rares sur des profils qui retrouveront un emploi rapidement. Un mauvais aiguillage a un coût humain (perte de chance pour l'usager) et un coût financier pour la collectivité (allocations prolongées, accompagnement mal calibré).

C'est ce qui motive le recours à l'IA : disposer d'un outil d'**aide à la décision**, exploitant à la fois les données structurées du dossier et le texte libre de l'entretien, capable de proposer en quelques secondes une estimation objectivée et reproductible du délai probable de retour à l'emploi, en complément — et non en remplacement — du jugement du conseiller.*

### 1.2 Énoncé du problème IA

*[Reformule en termes de tâche IA : « prédire X à partir de Y », « classer Z en N catégories », etc.]*

**Prédire, pour un usager donné, sa classe de délai de retour à l'emploi à partir des informations recueillies lors du premier entretien d'aiguillage.**

Reformulé en termes de tâche IA :

> **Classer** chaque demandeur d'emploi **en 3 catégories ordonnées** — 0 (retour rapide, < 6 mois), 1 (retour moyen, 6-12 mois), 2 (risque de longue durée, > 12 mois) — **à partir de** :
> - **X_tabulaire** : données administratives et socio-professionnelles structurées (âge, niveau de diplôme, ancienneté dans le dernier poste, code métier visé, code INSEE de la commune, statut d'allocataire, nationalité hors UE) ;
> - **X_texte** : la synthèse libre rédigée par le conseiller à l'issue de l'entretien (vectorisée via TF-IDF ou embeddings).

Il s'agit donc d'une **tâche supervisée de classification multimodale à 3 classes**, avec :
- une **cible catégorielle ordinale** (les 3 classes traduisent une gravité croissante du risque, ce qui a une incidence directe sur le choix des métriques et la gestion des erreurs) ;
- des **features hétérogènes** (numériques, catégorielles, haute cardinalité géographique, texte libre) nécessitant un pipeline de preprocessing et de fusion multimodale ;
- une **contrainte métier forte sur le type d'erreur toléré** : minimiser en priorité les faux négatifs sur la classe 2 (cas graves classés à tort comme rapides), plus que l'erreur globale.

### 1.3 Bénéficiaires & impacts

*[Direct : qui utilise la solution ? Indirect : qui subit les conséquences des prédictions ?]*

**Bénéficiaires directs — qui utilise la solution ?**
- **Les conseillers emploi** : ils saisissent les informations de l'usager via l'interface (formulaire), obtiennent une prédiction du niveau de risque en appui de leur diagnostic, et peuvent consulter l'historique des inférences. C'est l'utilisateur opérationnel de l'outil.
- **L'agence nationale de l'emploi** (pilotage/management) : dispose d'un outil de segmentation objectivée pour orienter l'allocation des dispositifs d'accompagnement et suivre les volumes par niveau de risque.

**Bénéficiaires indirects — qui subit les conséquences des prédictions ?**
- **Les demandeurs d'emploi eux-mêmes** : ce sont eux qui subissent concrètement l'issue de la prédiction — un usager classé en risque de longue durée devrait recevoir un accompagnement renforcé (bénéfice), tandis qu'un usager mal classé (faux négatif en classe 2 classé 0) risque d'être **privé de cet accompagnement indispensable**, avec un impact direct sur sa trajectoire de retour à l'emploi. Ils sont donc les plus exposés au risque en cas d'erreur, sans être les utilisateurs directs de l'outil.
- **La collectivité / les finances publiques** : impact indirect via le coût des allocations chômage prolongées et l'efficience globale des politiques d'accompagnement (un mauvais dimensionnement des ressources d'accompagnement a un coût financier).
- **L'administration elle-même sur le plan juridique** : en cas d'erreur d'orientation ayant causé une perte de chance avérée pour un usager, la responsabilité de l'éditeur ou de l'administration publique peut être engagée — un impact indirect mais réel de la qualité des prédictions.

Ce déséquilibre — celui qui *utilise* l'outil (le conseiller) n'est pas celui qui *subit* le plus fortement une erreur (l'usager) — est un point de vigilance éthique central du projet : il justifie que le conseiller garde la décision finale (l'IA reste une aide à la décision, pas une décision automatisée) et que les erreurs critiques (classe 2 → 0) soient traitées en priorité absolue dans l'optimisation du modèle.

### 3.2 Qualité des données : manquants, doublons, valeurs aberrantes

````python
# 3.2.1 Qualité des données
missing = pd.DataFrame({
    "Valeurs manquantes": df.isna().sum(),
    "Pourcentage (%)": (100 * df.isna().mean()).round(2).astype(str) + " %"
}).sort_values("Valeurs manquantes", ascending=False)

display(missing)

print("Nombre de lignes dupliquées (toutes colonnes) :", df.duplicated().sum())
print("Nombre d'identifiants usager dupliqués :", df["usager_id"].duplicated().sum())
````

````python
# 3.2.2 Détection des valeurs aberrantes (méthode IQR) sur les variables numériques
def detect_outliers_iqr(series, k=1.5):
    q1, q3 = series.quantile([0.25, 0.75])
    iqr = q3 - q1
    lower, upper = q1 - k * iqr, q3 + k * iqr
    outliers = series[(series < lower) | (series > upper)]
    return outliers, lower, upper

for col in ["age", "anciennete_poste_ans"]:
    outliers, lower, upper = detect_outliers_iqr(df[col].dropna())
    print(f"\n--- {col} ---")
    print(f"Bornes IQR : [{lower:.2f}, {upper:.2f}]")
    print(f"Nombre de valeurs aberrantes : {len(outliers)} ({len(outliers)/df[col].notna().sum()*100:.2f} %)")
    if len(outliers) > 0:
        print(outliers.sort_values(ascending=False).head(10))
````

📝 **Problèmes de qualité identifiés (§3.2)**

- Valeurs manquantes modérées sur 4 variables sur 10 (`age` ~4,9 %, `niveau_diplome` ~3,4 %,

  l'exploitabilité du dataset, mais nécessitant une stratégie d'imputation explicite en §4.
- Aucun doublon exact détecté, ni sur l'ensemble des colonnes ni sur `usager_id`.
- La détection IQR sur `age` et `anciennete_poste_ans` ne révèle pas d'outliers aberrants
  massifs — les valeurs extrêmes restent dans des plages plausibles (ex: 22,6 ans
  d'ancienneté), à confirmer visuellement par les boxplots ci-dessous plutôt qu'à traiter

````python
# Cartes de répartition géographique : département + région

# Préparation des variables géographiques ---

# Comme code_insee_commune contient généralement les 5 chiffres INSEE
#df["departement"] = df["code_insee_commune"].astype(str).str[:2]

# Correction DROM (codes à 3 chiffres tronqués par le str[:2])
is_drom = df["code_insee_commune"].astype(str).str.startswith("97")


df["region"] = df["departement"].map(region_map)
df.loc[is_drom, "region"] = "DROM"

# Effectifs par département et par région ---

dep_counts = df["departement"].value_counts().reset_index()
dep_counts.columns = ["code", "effectif"]

reg_counts = df["region"].value_counts().reset_index()
reg_counts.columns = ["region", "effectif"]
# Chargement du fond de carte départemental

# !pip install geopandas
france = gpd.read_file("../data/departements.geojson")

# Carte 1 : répartition par département (comme avant) ---

carte_dep = france.merge(dep_counts, left_on="code", right_on="code", how="left")
carte_dep["effectif"] = carte_dep["effectif"].fillna(0)

# Construction du fond de carte régional (dissolve département -> région) ---

france_reg = france.copy()
france_reg["region"] = france_reg["code"].map(region_map)
# Les DROM du geojson (971, 972, 973, 974, 976...) ne seront pas dans region_map
# (car region_map est indexé sur les codes métropolitains à 2 caractères) :
france_reg.loc[france_reg["code"].astype(str).str.startswith("97"), "region"] = "DROM"

# Fusion des polygones de département en un seul polygone par région
france_reg = france_reg.dissolve(by="region", as_index=False)

carte_reg = france_reg.merge(reg_counts, on="region", how="left")
carte_reg["effectif"] = carte_reg["effectif"].fillna(0)

# Affichage : département vs région ---

fig, axes = plt.subplots(1, 2, figsize=(18, 6))

carte_dep.plot(column="effectif", cmap="Blues", legend=True, ax=axes[0],
               edgecolor="grey", linewidth=0.3)
axes[0].set_title("Répartition géographique des bénéficiaires\npar département (96 modalités)")
axes[0].axis("off")

carte_reg.plot(column="effectif", cmap="Blues", legend=True, ax=axes[1],
               edgecolor="black", linewidth=0.6)
axes[1].set_title("Répartition géographique des bénéficiaires\npar région (14 modalités)")
axes[1].axis("off")

plt.tight_layout()
plt.show()
````

````python
# Vérification empirique du biais lexical suspecté sur synthese_entretien (cf. §2.4)
texte = df["synthese_entretien"].fillna("").str.lower()

# Recherche de mentions explicites de critères sensibles
patterns = {
    "age_mention": r"\b(?:\d+\s*ans|jeune|âgé|senior|ans\b)",
    "origine_mention": r"\b(?:étranger|immigr|nationalité|origine)",
    "genre_mention": r"\b(?:madame|monsieur)",
}
for label, pattern in patterns.items():
    n = texte.str.contains(pattern, regex=True, na=False).sum()
    print(f"{label} : {n} occurrences sur {texte.notna().sum()} textes ({n/len(df)*100:.1f} %)")

# Le texte permet-il de prédire une variable sensible ? (signal de biais lexical implicite)
tfidf = TfidfVectorizer(max_features=200, min_df=5)
X_text = tfidf.fit_transform(texte)

for sensible_col in ["nationalite_hors_ue", "groupe_age"]:
    y = df[sensible_col]
    mask = y.notna()
    if y[mask].nunique() < 2:
        continue
    clf = LogisticRegression(max_iter=1000)  # multi_class retiré (déprécié, comportement par défaut = multinomial)
    scoring = "roc_auc_ovr" if y[mask].nunique() > 2 else "roc_auc"
    scores = cross_val_score(clf, X_text[mask.values], y[mask], cv=5, scoring=scoring)
    print(f"\nAUC texte → {sensible_col} : {scores.mean():.3f} (proche de 0.5 = pas de biais lexical détecté)")
````

📝 **Analyse du texte libre `synthese_entretien`**

**Doublons stricts : 2410 sur 2500 lignes**
Le nombre de doublons est un premier signal fort : **96,4 % des synthèses sont des répétitions exactes** d'une formulation déjà vue ailleurs dans le dataset. Ce n'est pas un chiffre anodin pour du texte censé être rédigé librement par un conseiller — un tel taux de répétition serait extrêmement improbable sur de vraies notes d'entretien, où chaque conseiller formule les choses à sa manière.

**Confirmation formelle : seulement 9 formulations distinctes**
Le calcul explicite tranche la question : sur 2419 commentaires non vides, il n'existe que **9 valeurs uniques**, chacune répétée entre 202 et 338 fois. `synthese_entretien` n'est donc **pas du texte libre au sens propre** — c'est une variable catégorielle à 9 modalités, déguisée en champ texte. Cette découverte doit orienter tout le traitement en aval : un encodage catégoriel simple (one-hot) sera plus adapté et plus fiable qu'un pipeline NLP complet, qui n'apporterait aucune valeur ajoutée sur seulement 9 formulations fixes et risquerait même de perdre de l'information .

**Longueur du texte (63 à 77 caractères, 8-9 mots en moyenne)**
La distribution de longueur en caractères n'est pas continue mais **multimodale**, avec des pics nets à intervalles réguliers — cohérent avec 9 phrases de longueurs fixes qui se répètent, plutôt qu'une variation naturelle propre à du texte librement rédigé. Le nombre de mots est extrêmement resserré (écart-type de seulement 2 mots, plage de 0 à 12), ce qui confirme une fois de plus le caractère templatisé du champ : de vraies notes de conseiller varieraient bien davantage en longueur selon la complexité de chaque situation.

**Absence de biais lexical explicite ou implicite**
Deux vérifications convergent vers une conclusion rassurante sur le plan éthique :
- **Mentions explicites** de critères sensibles (âge, origine, genre) : **0 occurrence** sur les 2500 textes, quelle que soit la catégorie recherchée.
- **Test de prédictibilité indirecte** (AUC d'un classifieur texte → variable sensible) : **0,542** pour la nationalité, **0,543** pour le groupe d'âge — des scores quasiment identiques au hasard (0,5), ce qui indique que le vocabulaire des 9 formulations ne permet pas de reconstituer, même indirectement, l'origine ou l'âge de l'usager.

#### 3.5.4 Croisements clés : catégorielles vs cible

````python
# Croisements clés : catégorielles vs cible (tableaux croisés + test du Chi²) ---
pd.set_option('display.max_rows', None)  # évite toute troncature de l'affichage (ex. region à 14 modalités)

cat_vars = ['niveau_diplome', 'nationalite_hors_ue', 'est_allocataire' , 'synthese_entretien' , 'departement', 'region',
            'code_rome_vise', 'code_rome_famille']

n_vars = len(cat_vars)
n_cols = 2
n_rows = -(-n_vars // n_cols)  # arrondi supérieur → 4 lignes pour 8 variables

fig, axes = plt.subplots(n_rows, n_cols, figsize=(16, 4 * n_rows))
axes = axes.flatten()

resultats_chi2 = {}  # pour réutilisation éventuelle (tableau de synthèse, rapport)

for i, col in enumerate(cat_vars):
    ct_full = pd.crosstab(df[col], df['classe_retour_emploi'], normalize='index')

    # Test du Chi² sur la variable complète (jamais tronquée pour le calcul)
    ct_raw = pd.crosstab(df[col], df['classe_retour_emploi'])
    chi2, p, dof, expected = chi2_contingency(ct_raw)
    resultats_chi2[col] = {"chi2": chi2, "p": p, "ct": ct_full}

    print(f"\n--- {col} vs classe_retour_emploi (proportions par ligne) ---")
    print(ct_full.round(3).to_string())
    print(f"Chi² = {chi2:.2f}, p-value = {p:.4f}  (n modalités = {ct_full.shape[0]})")

    # --- Cas particulier : departement (96 modalités) → limiter au Top 20 par effectif ---
    if col == "departement":
        top_n = df[col].value_counts().head(20).index
        ct = ct_full.loc[top_n]
        titre_extra = " (Top 20 par effectif)"
        rotation, ha = 90, "center"

    # --- Cas particulier : code_rome_vise (50 modalités) → limiter au Top 15 pour l'affichage ---
    elif col == "code_rome_vise":
        top_n = df[col].value_counts().head(15).index
        ct = ct_full.loc[top_n]
        titre_extra = " (Top 15 par effectif)"
        rotation, ha = 90, "center"

    # --- Cas particulier : synthese_entretien (9 phrases longues) → labels raccourcis ---
    elif col == "synthese_entretien":
        ct = ct_full.copy()
        ct.index = [label[:22] + "…" if len(label) > 22 else label for label in ct.index]
        titre_extra = ""
        rotation, ha = 30, "right"

    else:
        ct = ct_full
        titre_extra = ""
        rotation, ha = 45, "right"

    ct.plot(kind='bar', stacked=True, colormap='viridis', ax=axes[i], legend=False)
    axes[i].set_title(f'{col} vs classe_retour_emploi{titre_extra}\n(Chi²={chi2:.1f}, p={p:.4f})',
                       fontsize=11)
    axes[i].set_ylabel('Proportion')
    axes[i].set_xlabel(col)
    axes[i].tick_params(axis='x', rotation=rotation)
    for label in axes[i].get_xticklabels():
        label.set_ha(ha)
        label.set_fontsize(8)

# Masque les sous-graphiques vides si besoin
for j in range(n_vars, len(axes)):
    axes[j].axis("off")

# Légende unique partagée
handles, labels = axes[0].get_legend_handles_labels()
fig.legend(handles, labels, title='classe_retour_emploi',
           loc='upper center', bbox_to_anchor=(0.5, 1.01), ncol=3)

plt.suptitle("Croisements variables catégorielles vs cible", fontsize=15, fontweight="bold", y=1.02)
plt.tight_layout()
plt.show()

pd.reset_option('display.max_rows')
````

#### 3.6.2 Disparate Impact

````python
# 2. Fonction de calcul du Disparate Impact (règle des 4/5) ---

def disparate_impact(df, sensitive_col, target_col, positive_value,
                              min_group_size=None):
    grp = df.groupby(sensitive_col, observed=True)[target_col]
    sizes = grp.size()
    selection_rate = grp.apply(lambda x: (x == positive_value).mean())

    if min_group_size is not None:
        keep = sizes[sizes >= min_group_size].index
        selection_rate = selection_rate.loc[keep]

    sr_min, sr_max = selection_rate.min(), selection_rate.max()
    di = sr_min / sr_max

    if di < 0.8:
        verdict = "⚠️ Signal"
    elif di > 1.25:
        verdict = "⚠️ Signal inversé"
    else:
        verdict = "✅ OK"

    return {
        "Variable": sensitive_col,
        "N groupes": len(selection_rate),
        "Selection Rate min": round(sr_min, 3),
        "Selection Rate max": round(sr_max, 3),
        "DI": round(di, 3),
        "Verdict": verdict
    }

# Calcul du DI pour chaque variable sensible + groupes de contrôle ---

sensitive_vars = [
    "nationalite_hors_ue",
           # remplace departement pour l'analyse d'équité
]

control_vars = [
    "groupe_age",
    "region",
]

results = []
for var in sensitive_vars + control_vars:
    results.append(
        disparate_impact(
            df=df,
            sensitive_col=var,
            target_col="classe_retour_emploi",
            positive_value=0  # classe 0 = retour rapide, considérée "issue favorable"
        )
    )

# Ajout du département brut à titre de comparaison (pour illustrer l'artefact)
#results.append(
#    disparate_impact(df, "departement", "classe_retour_emploi", 0)
#)
#results.append(
#    disparate_impact(df, "departement", "classe_retour_emploi", 0, min_group_size=15)
#)

di_table = pd.DataFrame(results).sort_values(by="DI").reset_index(drop=True)
di_table
````

📝 **Interprétation — Disparate Impact région (brut vs filtré)**

Le calcul brut du DI régional donne **0,689**, porté par le groupe **DROM (n=10,
SR=30,0 %)** — un effectif trop faible pour être statistiquement fiable (3 observations
en classe 0 sur 10 seulement). En excluant les groupes à effectif insuffisant
(`min_group_size=15`), le minimum se déplace sur la **Corse (n=21, SR=33,3 %)**, et le DI
remonte à **≈0,77** — toujours sous le seuil de 0,8, mais dans une zone d'alerte plus
modérée que le résultat brut ne le laissait penser.

**Conclusion** : la géographie (au niveau région) porte un signal d'inégalité réel mais
nettement moins sévère que l'âge (DI=0,245) ou la nationalité (DI=0,342), qui restent les
deux priorités du Scénario 2. Le résultat régional doit être rapporté avec sa réserve de
fiabilité (groupe DROM), plutôt que comme un chiffre unique et définitif.

#### 3.6.3 Analyse des variables proxy et fairness

Avant de construire le scénario S2, nous vérifions si des variables restantes peuvent contenir indirectement de l'information liée à nationalite_hors_ue.

**Principe :**

- nationalite_hors_ue reste disponible ici uniquement comme variable de référence pour l'audit ;
- elle sera supprimée des variables explicatives du scénario S2 ;
- une association statistique n'est pas une preuve de discrimination ;
- nous complétons l'analyse par un modèle auxiliaire permettant de mesurer la capacité des variables candidates à prédire la variable sensible.

Variables examinées : géographie/INSEE, diplôme, ROME, allocation, âge, ancienneté et texte d'entretien.

````python
proxy_required = [
    "age", "niveau_diplome",
    "code_insee_commune"
]
missing_proxy_cols = [c for c in proxy_required if c not in df.columns]
assert not missing_proxy_cols, f"Colonnes manquantes : {missing_proxy_cols}"

print("Analyse proxy prête :", len(proxy_required), "variables contrôlées.")
````

````python
from scipy.stats import chi2_contingency
categorical_proxy_vars = [
    "age",
    "niveau_diplome",
    "code_insee_commune"
]

proxy_chi2_results = []

for col in categorical_proxy_vars:
    tmp = df[[col, "nationalite_hors_ue"]].dropna()
    table = pd.crosstab(tmp[col], tmp["nationalite_hors_ue"])

    if table.shape[0] > 1 and table.shape[1] > 1:
        chi2, p_value, dof, expected = chi2_contingency(table)
        proxy_chi2_results.append({
            "variable": col,
            "n_modalites": table.shape[0],
            "chi2": chi2,
            "p_value": p_value
        })

proxy_chi2_results = pd.DataFrame(proxy_chi2_results).sort_values("p_value")
display(proxy_chi2_results)
````

**Test de proxy par modèle auxiliaire**
Nous construisons temporairement un modèle dont la cible est nationalite_hors_ue.

Une AUC nettement supérieure à 0,50 signifie que les variables candidates contiennent un signal permettant de retrouver indirectement la variable sensible. C'est un signal de proxy potentiel, pas une preuve de discrimination.

````python
# Proxy tabulaire : variables candidates -> nationalite_hors_ue
X_proxy = df[
    [
        "age",
        "niveau_diplome",
        "departement",
    ]
].copy()

y_proxy = df["nationalite_hors_ue"].copy()

num_proxy_cols = ["age"]
cat_proxy_cols = [
    "niveau_diplome",
    "departement"
]

proxy_preprocessor = ColumnTransformer([
    ("num", SimpleImputer(strategy="median"), num_proxy_cols),
    (
        "cat",
        Pipeline([
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("onehot", OneHotEncoder(handle_unknown="ignore"))
        ]),
        cat_proxy_cols
    )
])

proxy_model = Pipeline([
    ("preprocessing", proxy_preprocessor),
    ("model", RandomForestClassifier(
        n_estimators=300,
        random_state=42,
        class_weight="balanced",
        n_jobs=-1
    ))
])

proxy_cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

proxy_auc_scores = cross_val_score(
    proxy_model,
    X_proxy,
    y_proxy,
    cv=proxy_cv,
    scoring="roc_auc",
    n_jobs=-1
)

print(f"AUC proxy tabulaire : {proxy_auc_scores.mean():.3f} ± {proxy_auc_scores.std():.3f}")
````

````python
# Proxy texte : synthese_entretien -> nationalite_hors_ue
X_text_proxy = df["synthese_entretien"].fillna("").astype(str)
y_text_proxy = df["nationalite_hors_ue"]

text_proxy_model = Pipeline([
    ("tfidf", TfidfVectorizer(
        lowercase=True,
        ngram_range=(1, 2),
        min_df=2
    )),
    ("model", LogisticRegression(
        max_iter=1000,
        class_weight="balanced"
    ))
])

text_proxy_auc_scores = cross_val_score(
    text_proxy_model,
    X_text_proxy,
    y_text_proxy,
    cv=proxy_cv,
    scoring="roc_auc",
    n_jobs=-1
)

print(f"AUC proxy texte : {text_proxy_auc_scores.mean():.3f} ± {text_proxy_auc_scores.std():.3f}")
````

````python
sensitive_terms = [
    "nationalité", "nationalite",
    "origine", "pays",
    "langue", "linguistique",
    "immigration", "immigré", "immigre",
    "titre de séjour", "titre de sejour"
]

text_series = df["synthese_entretien"].fillna("").astype(str).str.lower()

term_counts = pd.DataFrame([
    {
        "terme": term,
        "occurrences": int(text_series.str.contains(term, regex=False).sum())
    }
    for term in sensitive_terms
]).sort_values("occurrences", ascending=False)

display(term_counts)
````

````python
# Décision opérationnelle issue de l'analyse proxy
tab_auc_mean = float(np.mean(proxy_auc_scores))
text_auc_mean = float(np.mean(text_proxy_auc_scores))

department_p = np.nan
if "departement" in proxy_chi2_results["variable"].values:
    department_p = float(
        proxy_chi2_results.loc[
            proxy_chi2_results["variable"].eq("departement"), "p_value"
        ].iloc[0]
    )

sensitive_text_hits = int(term_counts["occurrences"].sum())

# Seuil exploratoire : AUC >= 0.60 = signal prédictif notable.
# Il ne s'agit pas d'un seuil réglementaire ni d'une preuve de discrimination.
proxy_flags = []

if tab_auc_mean >= 0.60:
    proxy_flags.append("signal tabulaire")
if text_auc_mean >= 0.60:
    proxy_flags.append("signal texte")
if (
    not np.isnan(department_p)
    and department_p < 0.05
    and tab_auc_mean >= 0.60
):
    proxy_flags.append("signal géographique")

# Mitigation automatique uniquement si le signal est à la fois substantiel
# et cohérent avec l'analyse statistique.
PROXY_MITIGATION_FEATURES = []
if (
    tab_auc_mean >= 0.60
    and not np.isnan(department_p)
    and department_p < 0.05
):
    PROXY_MITIGATION_FEATURES.append("region")

proxy_risk_level = (
    "À surveiller / mitigation possible"
    if proxy_flags
    else "Faible à modéré — pas de mitigation automatique"
)

print(f"AUC proxy tabulaire : {tab_auc_mean:.3f}")
print(f"AUC proxy texte      : {text_auc_mean:.3f}")
print(
    f"p-value département  : {department_p:.4f}"
    if not np.isnan(department_p)
    else "p-value département : non disponible"
)
print(f"Occurrences de termes sensibles dans le texte : {sensitive_text_hits}")
print("Signal(s) retenu(s) :", proxy_flags or "aucun signal substantiel")
print("Niveau de risque proxy :", proxy_risk_level)
print("Mitigation appliquée à S2 :", PROXY_MITIGATION_FEATURES or "aucune")

if not proxy_flags:
    print("\nConclusion opérationnelle :")
    print("- les signaux de proxy sont faibles/modérés ;")
    print("- S2 retire la variable sensible explicite ;")
    print("- les variables métier restantes sont conservées ;")
    print("- la fairness réelle sera vérifiée après entraînement.")
else:
    print("\nConclusion opérationnelle : un signal proxy substantiel a été détecté.")
    print("S2 applique donc la mitigation définie ci-dessus, puis l'impact est contrôlé sur les performances et la fairness.")
````

````python
### a supprimer
# 4.2 Préprocesseur réutilisable

NUM_FEATURES = ["age", "anciennete_poste_ans"]
BASE_CAT_FEATURES = ["niveau_diplome", "code_rome_famille", "region", "est_allocataire"]
SENSITIVE_FEATURE = "nationalite_hors_ue"
TEXT_FEATURE = "synthese_entretien"

# Décision de mitigation issue de l'analyse proxy précédente.
MITIGATION_CAT_FEATURES = [c for c in PROXY_MITIGATION_FEATURES if c in BASE_CAT_FEATURES]

def make_preprocessor(num_features, cat_features, use_text):
    transformers = []

    if num_features:
        num_pipe = Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("scaler", StandardScaler())
        ])
        transformers.append(("num", num_pipe, num_features))

    if cat_features:
        cat_pipe = Pipeline([
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("onehot", OneHotEncoder(handle_unknown="ignore"))
        ])
        transformers.append(("cat", cat_pipe, cat_features))

    if use_text:
        text_pipe = Pipeline([
            ("imputer", SimpleImputer(strategy="constant", fill_value="")),
            ("flatten", FunctionTransformer(lambda x: x.ravel(), validate=False)),
            ("tfidf", TfidfVectorizer(
                lowercase=True,
                strip_accents="unicode",
                ngram_range=(1, 2),
                min_df=2,
                max_features=3000
            ))
        ])
        transformers.append(("text", text_pipe, [TEXT_FEATURE]))

    if not transformers:
        raise ValueError("Aucune feature sélectionnée")

    return ColumnTransformer(transformers, remainder="drop")

scenarios = {
    "S1": {
        "num": NUM_FEATURES,
        "cat": BASE_CAT_FEATURES + [SENSITIVE_FEATURE],
        "text": True,
        "description": "Multimodal complet"
    },
    "S2": {
        "num": NUM_FEATURES,
        "cat": [c for c in BASE_CAT_FEATURES if c not in MITIGATION_CAT_FEATURES],
        "text": True,
        "description": "Sans variable sensible + mitigation des proxies identifiés"
    },
    "S3": {
        "num": [],
        "cat": [],
        "text": True,
        "description": "Texte seul (TF-IDF)"
    },
    "S4": {
        "num": NUM_FEATURES,
        "cat": ["niveau_diplome"] + ["region"],
        "text": False,
        "description": "Tabulaire seul"
    }
}

print("Variables retirées de S2 au titre de la mitigation proxy :", MITIGATION_CAT_FEATURES or "Aucune")
pd.DataFrame(scenarios).T
````

### 4.3 Scénarios de jeux de données

*[Pour la certif, anticipe : 2 à 4 scénarios = avec/sans variables sensibles, avec/sans certaines variables corrélées à la cible. Liste-les ici, tu les compareras en section 6.]*

| Scénario | Variables conservées | Justification |
|---|---|---|
| S1 | Toutes | Baseline |
| S2 | Sans variables sensibles | Atténuation biais |
| S3 | … | … |

> ⚠️ **Un scénario change le périmètre, jamais la recette.**
> Chaque scénario instancie `make_preprocessor(...)` (4.2) avec ce qui lui
> reste : mêmes imputer / scaler / encodage partout, seul le périmètre change.
> C'est ce que veut dire « même prétraitement » dans la règle d'or de
> comparabilité — pas « le même objet `ColumnTransformer` ».
>
> Un scénario peut retirer **un type de données entier** et pas seulement
> quelques colonnes : d'où les blocs optionnels de `make_preprocessor`. Le
> nombre de branches du `ColumnTransformer` varie, les transformations non.
>
> Et on ne fait varier **qu'un facteur à la fois** : même modèle d'un scénario
> à l'autre, même scénario d'un modèle à l'autre. Sinon le tableau de la
> section 6 ne permet aucune conclusion.
>
> 💡 Si un scénario repose sur du *feature engineering* (variable dérivée,
> regroupement, agrégat), la colonne dérivée se calcule **avant**
> `make_preprocessor` et y entre comme une colonne ordinaire.

````python
import pandas as pd

feature_names = preprocessor.get_feature_names_out()

print("=" * 80)
print(" RAPPORT DÉTAILLÉ DU PRÉPROCESSOR & OPTIONS")
print("=" * 80)

for name, transformer, input_cols in preprocessor.transformers_:
    if name == 'remainder':
        continue
    
    block_features = [col for col in feature_names if col.startswith(f"{name}__")]
    
    print(f"\n🔹 Bloc : {name.upper()}")
    print(f"   • Colonnes d'entrée ({len(input_cols)}) : {list(input_cols)}")
    print(f"   • Total colonnes en sortie : {len(block_features)}")
    
    # 1. Détails spécifiques pour l'ORDINAL (Afficher le mapping exact)
    if name == 'ord':
        sub_pipeline = preprocessor.named_transformers_[name]
        ordinal_enc = sub_pipeline.named_steps['ordinal']
        for col, cats in zip(input_cols, ordinal_enc.categories_):
            mapping = {cat: idx for idx, cat in enumerate(cats)}
            print(f"     - Mapping défini pour '{col}' :")
            for cat, idx in mapping.items():
                print(f"         '{cat}' ---> {idx}")
            print(f"       * Valeurs inconnues / manquantes : encodées à {ordinal_enc.unknown_value}")

    # 2. Détails spécifiques pour le CATEGORICAL (One-Hot par variable d'origine)
    elif name == 'cat':
        print(f"     - Ventilation par colonne d'origine :")
        for orig_col in input_cols:
            # Filtrer les colonnes générées pour cette variable spécifique
            col_features = [f for f in block_features if f.startswith(f"cat__{orig_col}_")]
            print(f"         '{orig_col}' ---> {len(col_features)} colonnes binaires générées")

    # 3. Détails spécifiques pour le NUMERICAL
    elif name == 'num':
        sub_pipeline = preprocessor.named_transformers_[name]
        imputer = sub_pipeline.named_steps['imputer']
        print(f"     - Stratégie d'imputation : {imputer.strategy} (médiane)")
        print(f"     - Standardisation : StandardScaler actif (moyenne=0, variance=1)")

    # 4. Détails spécifiques pour le TEXT
    elif name == 'text':
        sub_pipeline = preprocessor.named_transformers_[name]
        tfidf = sub_pipeline.named_steps['tfidf']
        print(f"     - Options TF-IDF : max_features={tfidf.max_features}, ngram_range={tfidf.ngram_range}, min_df={tfidf.min_df}")
        print(f"     - Exemples de termes générés : {block_features[:5]}")

print("\n" + "=" * 80)
````

### 5.5 Optimisation des hyperparamètres

🎯 Démontrer que tu as **conscience** de l'effet des hyperparamètres et que tu as testé plusieurs configurations.

## Utiliser `config.py`, `preprocess.py` et `train.py`

Le flux d'entraînement du projet est centralisé dans `train.py`. Le notebook ne reconstruit donc pas les scénarios ni les pipelines : il lit la configuration, charge le scénario demandé et appelle `build_training_pipeline`, qui utilise lui-même le préprocesseur défini dans `preprocess.py`.

````python
# Les modules de src utilisent des imports internes de type `from config import ...`.
# On ajoute donc src au chemin Python avant d'importer train.py et config.py.
SRC_ROOT = PROJECT_ROOT / "src"
if str(SRC_ROOT) not in sys.path:
    sys.path.insert(0, str(SRC_ROOT))

import config
import train as train_module

MODEL_TYPE = "logistic_regression"
SCENARIO_NAME = "multimodal_ethique"
CONFIG_NAME = "balanced"

print("Dataset :", config.DATA_PATH)
print("Scénario :", SCENARIO_NAME)
print("Paramètres du scénario :", config.SCENARIOS[SCENARIO_NAME])
print("Configuration modèle :", CONFIG_NAME)
````

````python
from sklearn.metrics import accuracy_score, f1_score

# Chargement du scénario via train.py : la fonction délègue à preprocess.py.
X_scenario, y_scenario = train_module.load_scenario_dataset(
    data_path=config.DATA_PATH,
    scenario_name=SCENARIO_NAME,
)

X_train_scenario, X_holdout_scenario, y_train_scenario, y_holdout_scenario = train_test_split(
    X_scenario,
    y_scenario,
    test_size=config.TEST_SIZE,
    stratify=y_scenario,
    random_state=config.RANDOM_STATE,
)

pipeline = train_module.build_training_pipeline(
    model_type=MODEL_TYPE,
    scenario_name=SCENARIO_NAME,
    config_name=CONFIG_NAME,
)

pipeline.fit(X_train_scenario, y_train_scenario)
y_pred_scenario = pipeline.predict(X_holdout_scenario)

print("Train :", X_train_scenario.shape)
print("Holdout :", X_holdout_scenario.shape)
print("Accuracy holdout :", round(accuracy_score(y_holdout_scenario, y_pred_scenario), 4))
print("F1 macro holdout :", round(f1_score(y_holdout_scenario, y_pred_scenario, average="macro"), 4))
print("Pipeline :", pipeline.named_steps.keys())
````
